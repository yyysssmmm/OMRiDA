```python
# 피클 담구기
import os
import pickle
from PIL import Image
import numpy as np

image_dir = "/content/drive/MyDrive/AI_Project/TAMER/data/cropme/2019"       # printed 이미지 폴더
caption_path = "/content/drive/MyDrive/AI_Project/TAMER/data/crohme/2019/caption.txt"       # caption.txt 파일
output_pkl = "/content/drive/MyDrive/AI_Project/TAMER/data/cropme/2019_images.pkl"   # 테스트용 출력 파일

# 📑 caption.txt 기준으로 이미지 이름 리스트 확보
with open(caption_path, "r", encoding="utf-8") as f:
    lines = f.readlines()

images_dict = {}
missing_count = 0

for line in lines:
    tokens = line.strip().split()
    if not tokens:
        continue

    img_key = tokens[0]                     # 예: RIT_2014_293
    filename = img_key + ".png"             # printed 이미지는 확장자 있는 파일
    img_path = os.path.join(image_dir, filename)

    if not os.path.exists(img_path):
        missing_count += 1
        continue

    # 흑백 이미지로 로딩
    img = Image.open(img_path).convert("L")
    img_np = np.array(img).astype(np.uint8)

    # 크기 제한 (320000 초과시 리사이즈)
    if img_np.shape[0] * img_np.shape[1] > 320000:
        base_height = 256
        h_percent = base_height / img_np.shape[0]
        w_size = int(img_np.shape[1] * h_percent)
        img = img.resize((w_size, base_height), Image.ANTIALIAS)
        img_np = np.array(img).astype(np.uint8)

    images_dict[img_key] = img_np

# ✅ 저장
with open(output_pkl, "wb") as f:
    pickle.dump(images_dict, f)

print(f"[✔] images.pkl saved with {len(images_dict)} images to: {output_pkl}")
print(f"[ℹ️] Number of missed images: {missing_count}")
```
